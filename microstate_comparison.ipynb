```python
import mne
import os
```

```python
data_folder = "data"

files = [f for f in os.listdir(data_folder) if f.endswith(".set")]

print(files[0])
```

```python
file_path = os.path.join(data_folder, files[0])

raw = mne.io.read_raw_eeglab(file_path, preload=True)

print(raw)
```

```python
data = raw.get_data()

print(data.shape)
```

```python
import sys
sys.path.append("code")

from microstates import segment_microstates
```

```python
methods = ["mod_kmeans", "AAHC", "TAAHC"]
k_values = [4,5,6,7]
```

```python
results = {}

for method in methods:
    
    results[method] = {}
    
    for k in k_values:
        
        result = segment_microstates(
            data,
            method=method,
            n_states=k
        )
        
        maps = result[0]
        
        results[method][k] = maps

print("Microstate segmentation finished")
```
